### Travelling Salesman Problem (TSP) Implementation
The Travelling Salesman Problem asks: Given a list of cities and the distances between each pair of cities, what is the shortest possible route that visits each city exactly once and returns to the origin city?

Below is the **Held-Karp algorithm** implementation, a dynamic programming approach that solves TSP in $O(2^n \cdot n^2)$ time complexity, which is significantly faster than the $O(n!)$ brute-force method.

```python
import math

def tsp_held_karp(distance_matrix):
    """
    Solves the Travelling Salesman Problem using the Held-Karp dynamic programming algorithm.
    :param distance_matrix: 2D list where distance_matrix[i][j] is the distance from city i to city j
    :return: A tuple of (minimum_distance, path)
    """
    n = len(distance_matrix)
    memo = {}

    def solve(mask, u):
        # Base case: All cities have been visited
        if mask == (1 << n) - 1:
            return distance_matrix[u][0], [0]  # Return to the starting city (0)

        state = (mask, u)
        if state in memo:
            return memo[state]

        min_dist = math.inf
        best_path = []

        for v in range(n):
            # If city v has not been visited yet
            if not (mask & (1 << v)):
                new_mask = mask | (1 << v)
                dist, path = solve(new_mask, v)
                total_dist = distance_matrix[u][v] + dist
                if total_dist < min_dist:
                    min_dist = total_dist
                    best_path = [v] + path

        memo[state] = (min_dist, best_path)
        return memo[state]

    # Start from city 0 with only city 0 visited (mask = 1)
    min_distance, path = solve(1, 0)
    full_path = [0] + path

    return min_distance, full_path

# --- Example Graph Setup ---
# Distance matrix for 4 cities (0, 1, 2, 3)
tsp_matrix = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

min_cost, optimal_path = tsp_held_karp(tsp_matrix)

print(f"Minimum Travelling Cost: {min_cost}")
print("Optimal Tour Path:", " -> ".join(map(str, optimal_path)))
```
